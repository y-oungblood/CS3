## 1. Artifact sanity check

Neighbors come from LensKit's `ItemKNNScorer` (cosine on item-mean-centered ratings). *Done when* well-known movies have sensible neighbors.

```python
import json, sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'offline' else Path.cwd()
sys.path.insert(0, str(ROOT / 'offline'))
from show_neighbors import neighbors_for

ART = ROOT / 'data' / 'artifacts' / 'deploy'
movies = pd.read_parquet(ART / 'movies.parquet')
nbrs = pd.read_parquet(ART / 'neighbors.parquet')
json.loads((ART / 'meta.json').read_text())
```

```python
for title in ['The Dark Knight', 'Toy Story', 'Pulp Fiction', 'Arrival', 'The Notebook', 'Spirited Away']:
    try:
        t = neighbors_for(title, movies, nbrs, 8)
    except KeyError:
        print('not found:', title); continue
    print('
' + t.attrs['movie']); display(t)
```
